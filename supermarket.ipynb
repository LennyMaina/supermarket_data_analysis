```python
transaction_query = '''
select * from supermarket.transaction db;
'''
```

```python
transaction_df = pd.read_sql(transaction_query, con = engine)
```

Dropping duplicates

```python
transaction_df = transaction_df.drop_duplicates()
```

```python
purchase_df = purchase_df.drop_duplicates()
```

Checking data types

```python
transaction_df.info()
```

Extracting year and month

```python
transaction_df['YEAR'] = transaction_df['DATE'].dt.year
```

```python
transaction_df.head()
```

```python
transaction_df['MONTH'] = transaction_df['DATE'].dt.month
```

```python
transaction_df.head()
```

```python
merged_df = transaction_df.merge(purchase_df, on = 'LYLTY_CARD_NBR', how='left')
```

```python
merged_df.head()
```

```python
merged_df.isna().sum()
```

Top sold products

```python
top_products = (merged_df.groupby('PROD_NAME')['PROD_QTY'].sum().sort_values(ascending = False).head())
```

```python
top_products.head()
```

Total sales by lifestage

```python
sales_by_lifestage = (merged_df.groupby('LIFESTAGE')['TOT_SALES']).sum().sort_values(ascending = False)
```

```python
sales_by_lifestage.head()
```

Time series analysis

Monthly sales each year

```python
merged_df['MONTH_NAME'] = merged_df['DATE'].dt.month_name()
```

```python
merged_df.head()
```

```python
monthly_sales = (merged_df.groupby(['YEAR', 'MONTH_NAME'])['TOT_SALES'].sum().sort_values(ascending = False))
```

```python
monthly_sales
```

Store perfomance
1. Total sales by store

```python
store_sales = (
    merged_df.groupby('STORE_NBR')
    ['TOT_SALES']
    .sum()
    .sort_values(ascending = False)
    )
```

```python
store_sales
```

Product perfomance

```python
top_products = (merged_df.groupby('PROD_NAME')['TOT_SALES'].sum().sort_values(ascending = False))
```

```python
top_products.reset_index().head()
```
